## 第二步：一日游 vs 趋势性判断

## 第二步：板块内资金的「攻击结构」诊断

2. **市值偏好**：流入标的是大市值龙头还是小市值弹性标的？

   - 大小市值混搭 → 多路资金共振，最优信号

**板块关联分析**（即使 N 小也要做）：
- 少数上榜板块之间是否存在产业链/逻辑关联？
- 如果 N=4 且全部弱势（强度 <55），分析重点变为「为什么市场整体疲软，这些板块却进入了排名？」——是防御属性、逆周期、还是底部信号？

#### 1C. 形态 C（N=1）：板块定性

单个板块处理：
- 联网搜索该板块的定义、成分构成、核心行业覆盖
- **关键判断**：该板块是「窄基概念」（如培育钻石，单一产业链）还是「宽基概念」（如同花顺新质50，涵盖多个不相关赛道）？
- 如果是宽基概念 → 进入「筐内赛道分解」模式（见第三步·C）

---

### ═══ 第二步：个股层分析（框架随 M 变化）═══

#### 2A. 通用步骤（所有形态）

**行业分布统计**：联网搜索每只个股的核心行业归属（非概念标签），统计行业集中度：
- 哪些行业出现了 3 只以上个股？→ 形成个股层面的「行业集群」
- 个股行业集群是否与板块 Top 中的板块对应？→ 预判映射关系

**强度分级分布**：
- 极端强度（≥90）/ 高强度（80-89）/ 中等（65-79）/ 偏弱（≤64）
- 与板块强度分布做对比：个股极端强度比例 vs 板块极端强度比例

**涨跌面**：
- 上涨/下跌比例 + 涨幅-强度匹配度
- 低涨幅+高强度 → 蓄势型，标注「隐蔽信号」

**市值风格**：通过价格/名称/搜索粗略判断大盘/中盘/小盘分布，判断资金偏好。

#### 2B. 形态 B 特别关注

当 M ≤15（个股数量较少）时：
- 为什么是这些个股入围？矩阵模式（尤其是与板块不同的模式）如何影响了筛选结果？
- 个股的行业集中度比数量本身更重要——10 只中医药占 8 只的信号强度远超 20 只中分散在 10 个行业

#### 2C. 形态 C 特别关注

当 N=1 且 M≥10 时：
- 将个股按行业赛道重新聚类（半导体/AI/机器人/光通信/生物制造等）
- 统计各赛道的个股数量、平均强度、平均涨幅
- 这是「筐内赛道分化」分析的基础

---

### ═══ 第三步：板块-个股交叉验证（框架随形态变）═══

#### 3A. 形态 A 交叉验证：经典共振/背离矩阵

为每只个股建立映射记录：

| 股票 | 搜索确认的所属板块 | 该板块是否在 Top N？ | 板块强度 | 个股强度 | 映射信号 |
|:---|:---|:---|:---|:---|:---|

映射信号判定（四种）：
1. **🔴 完美共振**：板块在 Top N 且强度≥80 + 个股强度≥80 → 板块-个股双确认，信号最强
2. **🟠 板块驱动**：板块强度≥80 但个股强度<80 → 板块风口已起，个股可能滞后
3. **🔵 个股突围**：个股强度≥80 但板块不在 Top N 或板块强度<80 → 个股独立行情
4. **⚪ 待确认**：板块强度<80 且个股强度<80 → 需要额外催化验证

**共振/背离面板**（三个卡片并排）：
- **最强共振**：列出完美共振个股及所属板块集群，统计各板块集群的共振个股数（渗透度）
- **关键背离**：板块强/个股弱（集群空壳）、板块弱/个股强（个股突围）、板块和个股属于完全不同集群（双主线）
- **隐蔽信号**：涨幅小但强度高的板块和个股（蓄势型）

#### 3B. 形态 B 交叉验证：背离驱动分析

当板块数量稀少时，经典映射大部分个股会落入「🔵 个股突围」——这本身就是核心信号：

**核心诊断**：
- 个股的行业集群（如医药 8/10）是否出现在板块 Top 中？
  - 如果否 → 这是最关键的背离信号。板块筛的是低位拐头（右侧趋势），医药集群虽未整体拐头但个股已在突围 → 医药可能处于「板块未动、个股先行」的前夜
  - 如果是 → 判断共振质量

**关键问题**：
- 板块和个股的不同矩阵模式，如何导致了当前的数据格局？
- 双模式差异（如板块右侧趋势 vs 个股全量排名）本身就是一种「跨模式信号」

#### 3C. 形态 C 交叉验证：筐内赛道分解

当 N=1 时，不存在传统意义的「板块-个股映射」，分析框架切换为：

**筐内赛道分解**：
1. 将个股按行业赛道聚类（联网搜索确认每只所属行业）
2. 统计各赛道的：个股数量 + 平均强度 + 平均涨幅 + 强度极差
3. 判断「筐内主线」：哪个赛道在筐内最强？
4. 判断「筐内轮动」：强度极差大的赛道是否出现了龙头切换？

**板块概念 vs 个股实质分析**：
- 单板块如果是宽基概念（如「新质生产力」涵盖半导体+AI+生物制造+机器人），分析重点是筐内各赛道分化
- 单板块如果是窄基概念（如「培育钻石」），分析重点是筐内个股的产业链定位（上游设备/中游制造/下游品牌）

**不需要映射矩阵表**（因为所有个股都在同一个筐里），替换为「筐内赛道对比表」。
